```python
import os
import subprocess
import matplotlib.pyplot as plt
import numpy as np
#IUCN Red List Categories
Red_List = ["EX - Extinct", 
            "EW - Extinct In The Wild(only found in zoos)",
           "RE -Regionally Extinct", 
            "CR - Critically Endangered", 
            "EN -Endangered",
           "VU - Vulnerable", 
            "LR/cd - Lower Risk: Conservation Dependent", 
            "NT - Near Threatened",
           "LC - Least Concern", 
            "DD - Data Deficient", 
            "NA - Not Applicable"]

# Planted seedlings in the Alan’ny Varika Reserve:
Tree_Stats = {
    "Statistics Links": {
        "Pithecellobium dulce":{
            "stat_link": "https://www.iucnredlist.org/species/130519803/149057978",
            "extra_link": "https://www.cabidigitallibrary.org/doi/full/10.1079/cabicompendium.41187#abstract",
            "Conservation_status": Red_List[8],
            "Endemic":False },
        "Neotina isoneura":{
            "stat_link": "https://www.iucnredlist.org/species/137813882/137870748",
            "Conservation_status": Red_List[8],
            "Endemic":True },
        "Tamarindus indica":{
            "stat_link": "https://www.iucnredlist.org/species/62020997/62020999",
            "extra link": "https://www.cabidigitallibrary.org/doi/10.1079/cabicompendium.54073",
            "Conservation_status": Red_List[8],
            "Endemic":False },
        "Grewia androyensis":{ 
            "stat_link": "https://www.iucnredlist.org/species/128258055/128260099",
            "Conservation_status": Red_List[8],
            "Endemic":True },
        "Cordia caffra":{ 
            "stat_link": "https://www.iucnredlist.org/species/146302878/208442391",
            "Conservation_status": Red_List[8],
            "Endemic":False },
        "Celtis madagascariensis":{ 
            "stat_link": "https://www.iucnredlist.org/species/128086221/128090674",
            "Conservation_status": Red_List[8],
            "Endemic":False },
        "Rinorea greveana":{ 
            "stat_link": "https://www.iucnredlist.org/species/128659574/128660217",
            "Conservation_status": Red_List[8],
            "Endemic":True },
        "Strychnos madagascariensis":{ 
            "stat_link": "https://www.iucnredlist.org/species/146211218/146211220",
            "Conservation_status": Red_List[8],
            "Endemic":False },
        "Tricalysia dauphiniensis":{
            "stat_link": "https://www.iucnredlist.org/species/69222113/69236223",
            "Conservation_status": Red_List[5],
            "Endemic":True 
        }
    }
}

# Logic
def print_Tree_Stats(Tree_Stats, Red_List):
    for Species, details in Tree_Stats["Statistics Links"].items():
        print(f"{Species}: {details['Endemic']}")

# Uses a starting population of 1,000 to illustrate the point that the population loss
# is not that bad after the first two years. 
# I did not write this Graphs() code block...sry I aquiested to the chat bot
def Tree_Mortality():
    directory = "."
    os.chdir(directory)
    # Constants
    total_pop = 1000  # Starting population
    mortality_rates = {
        "Nursery": {"A": 0.60, "B": 0.35},  # Mortality rates in Nursery
        "Field": {"A": 0.20, "B": 0.15}    # Mortality rates in Field
    }
    time_ranges = {"Nursery": (0.25, 0.5), "Field": (0.5, 2)}

    # Linear decay function
    def compute_decay(base_population, mortality_rate, time_range, time):
        t_min, t_max = time_range
        fraction = (time - t_min) / (t_max - t_min)
        return base_population * (1 - mortality_rate * fraction)

    # Time ranges
    Nursery_Time = np.linspace(*time_ranges["Nursery"], 10)
    Field_Time = np.linspace(*time_ranges["Field"], 20)

    # Compute populations for Nursery
    Nursery_A_values = compute_decay(total_pop, mortality_rates["Nursery"]["A"], time_ranges["Nursery"], Nursery_Time)
    Nursery_B_values = compute_decay(total_pop, mortality_rates["Nursery"]["B"], time_ranges["Nursery"], Nursery_Time)

    # Compute populations for Field (start with ending populations of Nursery)
    final_A_Nursery = Nursery_A_values[-1]  # Population at the end of Nursery A
    final_B_Nursery = Nursery_B_values[-1]  # Population at the end of Nursery B

    Field_A_values = compute_decay(final_A_Nursery, mortality_rates["Field"]["A"], time_ranges["Field"], Field_Time)
    Field_B_values = compute_decay(final_B_Nursery, mortality_rates["Field"]["B"], time_ranges["Field"], Field_Time)

    # Combine populations for continuous plotting
    A_Time = np.concatenate((Nursery_Time, Field_Time))
    A_Population = np.concatenate((Nursery_A_values, Field_A_values))

    B_Time = np.concatenate((Nursery_Time, Field_Time))
    B_Population = np.concatenate((Nursery_B_values, Field_B_values))

    # Graph
    plt.figure(figsize=(10, 5))

    # Plot continuous lines for A and B
    plt.plot(A_Time, A_Population, label="Native Species", color='red')
    plt.plot(B_Time, B_Population, label="Non-Native Species", color='black')

    # Add dividing line
    plt.axvline(x=0.5, color='black', linestyle='--', label="Transition (0.5)")

    # Labels and formatting
    plt.xlabel("Time (in years)")
    plt.ylabel("Population")
    plt.title("Mortality Rate Decay (for tree species)")
    plt.legend()
    plt.grid()
    plt.savefig("Mortality Rate in Trees.pdf", format="pdf")
    plt.show()

import numpy as np
import matplotlib.pyplot as plt

import numpy as np
import matplotlib.pyplot as plt

def Tree_Growth():
    directory = "."
    os.chdir(directory)
    # Constants
    total_pop = 1000  # Starting population of each species
    # Realistic DBH growth rates (in cm per year)
    growth_rates = {
        "P.dulce": {"Low": 3.75, "High": 6.25},  # Average yearly growth over 8 years
        "Tamarindus indica": {"Low": 2.5, "High": 5.0}  # Average yearly growth over 12 years
    }
    time_ranges = {"P.dulce": (0, 8), "Tamarindus indica": (0, 12)}  # Growth periods in years

    # Linear growth function
    def compute_growth(initial_dbh, growth_rate, time):
        return initial_dbh + (growth_rate * time)

    # Time intervals
    Pdulce_Time = np.linspace(*time_ranges["P.dulce"], 100)  # 100 points over 8 years
    Tamarindus_Time = np.linspace(*time_ranges["Tamarindus indica"], 100)  # 100 points over 12 years

    # Compute DBH values for P. dulce
    Pdulce_growth = [
        compute_growth(0, growth_rate, Pdulce_Time) for growth_rate in growth_rates["P.dulce"].values()
    ]

    # Compute DBH values for Tamarindus indica
    Tamarindus_growth = [
        compute_growth(0, growth_rate, Tamarindus_Time) for growth_rate in growth_rates["Tamarindus indica"].values()
    ]

    # Visualization
    plt.figure(figsize=(10, 6))

    # Plot growth for P. dulce
    plt.plot(Pdulce_Time, Pdulce_growth[0], label="P. dulce (Low Growth)", color="green")
    plt.plot(Pdulce_Time, Pdulce_growth[1], label="P. dulce (High Growth)", color="darkgreen")

    # Plot growth for Tamarindus indica
    plt.plot(Tamarindus_Time, Tamarindus_growth[0], label="Tamarindus indica (Low Growth)", color="orange")
    plt.plot(Tamarindus_Time, Tamarindus_growth[1], label="Tamarindus indica (High Growth)", color="darkorange")

    # Labels and Formatting
    plt.xlabel("Time (in years)")
    plt.ylabel("DBH (in cm)")
    plt.title("Tree Species Growth in DBH Over Time")
    plt.legend()
    plt.grid()
    plt.savefig("Tree Species Growth in DBH Over Time.pdf", format="pdf")
    plt.show()






```
